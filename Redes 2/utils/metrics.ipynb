```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def print_results(total_tcp, per_package_tcp, total_udp, per_package_udp, lost_packages):
    total_tcp_mean, total_tcp_std = np.mean(total_tcp), np.std(total_tcp)
    per_package_tcp_mean, per_package_tcp_std = np.mean(per_package_tcp), np.std(per_package_tcp)

    total_udp_mean, total_udp_std = np.mean(total_udp), np.std(total_udp)
    per_package_udp_mean, per_package_udp_std = np.mean(per_package_udp), np.std(per_package_udp)
    lost_packages_mean, lost_packages_std = np.mean(lost_packages), np.std(lost_packages)

    print("========================================================")
    print(f"Total TCP Mean: {total_tcp_mean} - Total TCP Std: {total_tcp_std}")
    print(f"Per package TCP Mean: {per_package_tcp_mean} - Per package TCP Std: {per_package_tcp_std}\n")

    print(f"Total UDP Mean: {total_udp_mean} - Total UDP Std: {total_udp_std}")
    print(f"Per package UDP Mean: {per_package_udp_mean} - Per package UDP Std: {per_package_udp_std}")
    print(f"Lost packages Mean: {lost_packages_mean} - Lost packages Std: {lost_packages_std}")
    print("========================================================")
```

## Experimento 1

```python
total_tcp = [55.9395, 55.9410, 56.0252, 55.9363, 56.0266]
per_package_tcp = [0.005566, 0.005563, 0.005573, 0.005564, 0.005567]

total_udp = [54.7663, 54.7793, 54.8630, 54.7742, 54.8641]
per_package_udp = [0.005415, 0.005418, 0.005426, 0.005417, 0.005416]
lost_packages = [0, 0, 1, 1, 1]

print_results(total_tcp, per_package_tcp, total_udp, per_package_udp, lost_packages)
```

```python
values_tcp = [5.5947000000000005, 28.0873, 55.96733333333333, 83.96633333333334, 111.93333333333334, 139.95983333333334, 279.9538]
values_udp = [5.535933333333333, 27.384600000000002, 54.80036666666666, 82.21706666666667, 109.64546666666666, 137.04276666666667, 274.0776333333333]
packages_qty = [1000, 5000, 10000, 15000, 20000, 25000, 50000]

bar_width = 0.4
x = np.arange(len(packages_qty))

plt.figure(figsize=(8, 5))

plt.bar(x - bar_width / 2, values_tcp, width=bar_width, label="TCP", color='blue')
for i, y in enumerate(values_tcp):
    plt.text(x[i] - bar_width / 2, y + 2, f'{y:.2f}', fontsize=8, ha='center', color='black')

plt.bar(x + bar_width / 2, values_udp, width=bar_width, label="UDP", color='orange')
for i, y in enumerate(values_udp):
    plt.text(x[i] + bar_width / 2, y + 2, f'{y:.2f}', fontsize=8, ha='center', color='black')

plt.title("Tempo de Transmissão de Pacotes", fontsize=14)
plt.xlabel("Quantidade de Pacotes Enviados", fontsize=12)
plt.ylabel("Tempo Gasto (segundos)", fontsize=12)
plt.xticks(ticks=x, labels=[f"{q:,}" for q in packages_qty], fontsize=10) 
plt.legend(title="Protocolo", fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
